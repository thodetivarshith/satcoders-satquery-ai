```python
import subprocess

code = r'''
import sys
import os
import torch

repo = "/content/satcoders-satquery-ai/models/geochat_finetuned/GeoChat"
data = "/content/satcoders-satquery-ai/models/geochat_finetuned/conversations.json"
images = "/content/satcoders-satquery-ai/models/geochat_finetuned/images"

sys.path.insert(0, repo)

from transformers import BitsAndBytesConfig
from geochat.model.language_model.geochat_llama import GeoChatLlamaForCausalLM

print("Loading model...")

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
)

model = GeoChatLlamaForCausalLM.from_pretrained(
    "MBZUAI/geochat-7B",
    quantization_config=bnb_config,
    device_map="auto",
    low_cpu_mem_usage=True,
)

print("Initializing vision...")

class Args:
    vision_tower = "openai/clip-vit-large-patch14-336"
    mm_vision_select_layer = -2
    mm_projector_type = "mlp2x_gelu"
    mm_use_im_start_end = False
    mm_use_im_patch_token = False
    mm_vision_select_feature = "patch"
    pretrain_mm_mlp_adapter = None
    tune_mm_mlp_adapter = False

model_args = Args()

model.get_model().initialize_vision_modules(
    model_args=model_args,
    fsdp=None
)

vision_tower = model.get_vision_tower()
vision_tower.to(dtype=torch.float16, device=torch.device("cuda"))

print("Preparing LoRA...")

from peft import prepare_model_for_kbit_training, LoraConfig, get_peft_model

model.config.use_cache = False

model = prepare_model_for_kbit_training(
    model,
    use_gradient_checkpointing=True
)

model.enable_input_require_grads()

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "v_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
)

model = get_peft_model(model, lora_config)

print("Loading tokenizer...")

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "MBZUAI/geochat-7B",
    model_max_length=512,
    padding_side="right",
    use_fast=False,
)

tokenizer.pad_token = tokenizer.unk_token

print("Loading one training example...")

import json
from PIL import Image

with open(data, "r", encoding="utf-8") as f:
    records = json.load(f)

record = records[0]
image_path = os.path.join(images, record["image"])

image = Image.open(image_path).convert("RGB")

print("Example image:", record["image"])
print("Question:", record["conversations"][0]["value"])
print("Answer:", record["conversations"][1]["value"])
print("Image size:", image.size)

print("\nEnvironment prepared successfully.")
print("VRAM allocated (GB):",
      round(torch.cuda.memory_allocated() / 1024**3, 2))
'''

result = subprocess.run(
    ["/content/geochat-env/bin/python", "-c", code],
    capture_output=True,
    text=True
)

print("=== STDOUT ===")
print(result.stdout)

print("=== STDERR ===")
print(result.stderr)

print("=== EXIT CODE ===")
print(result.returncode)
```

```python
import subprocess
import os
import sys

repo = "/content/satcoders-satquery-ai/models/geochat_finetuned/GeoChat"
python = "/content/geochat-env/bin/python"

data_path = "/content/satcoders-satquery-ai/models/geochat_finetuned/conversations.json"
image_folder = "/content/satcoders-satquery-ai/models/geochat_finetuned/images"
output_dir = "/content/drive/MyDrive/satquery/checkpoints/geochat_v1_bigearth"

os.makedirs(output_dir, exist_ok=True)

cmd = [
    python,
    os.path.join(repo, "geochat", "train", "train_mem.py"),

    "--model_name_or_path", "MBZUAI/geochat-7B",
    "--version", "v1",

    "--data_path", data_path,
    "--image_folder", image_folder,

    "--vision_tower", "openai/clip-vit-large-patch14-336",
    "--mm_projector_type", "mlp2x_gelu",
    "--mm_vision_select_layer", "-2",
    "--mm_use_im_start_end", "False",
    "--mm_use_im_patch_token", "False",
    "--image_aspect_ratio", "pad",

    "--lora_enable", "True",
    "--bits", "4",

    "--fp16", "True",

    "--output_dir", output_dir,

    "--num_train_epochs", "1",
    "--per_device_train_batch_size", "1",
    "--per_device_eval_batch_size", "1",
    "--gradient_accumulation_steps", "16",

    "--learning_rate", "2e-4",
    "--weight_decay", "0",
    "--warmup_ratio", "0.03",
    "--lr_scheduler_type", "cosine",

    "--evaluation_strategy", "no",

    "--save_strategy", "steps",
    "--save_steps", "10",
    "--save_total_limit", "2",

    "--logging_steps", "1",
    "--report_to", "none",

    "--gradient_checkpointing", "True",
    "--lazy_preprocess", "True",
    "--model_max_length", "512",

    "--dataloader_num_workers", "2",
]

print("🚀 Starting full GeoChat fine-tuning...")
print("Dataset : 488 records")
print("Images  : 487")
print("GPU     : T4")
print("Output  :", output_dir)
print()

process = subprocess.Popen(
    cmd,
    cwd=repo,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)

for line in process.stdout:
    print(line, end="")

return_code = process.wait()

print()
if return_code == 0:
    print("✅ FULL GEochat fine-tuning completed successfully.")
else:
    print(f"❌ Training stopped with exit code: {return_code}")
```

```python
 import zipfile
from pathlib import Path

src = Path("geochat_dataset_487.zip")
dst = Path("geochat_dataset_487_clean.zip")

with zipfile.ZipFile(src, "r") as zin:
    with zipfile.ZipFile(dst, "w", zipfile.ZIP_DEFLATED) as zout:
        for item in zin.infolist():
            clean_name = item.filename.replace("\\", "/")
            data = zin.read(item.filename)
            zout.writestr(clean_name, data)

print("Created:", dst)
print("Size:", dst.stat().st_size, "bytes")
```
